```python
#........................ Student Performance analysis.......................
# 1. Introduction
#This project analyzes the academic performance of students using Python and Pandas.
#The dataset contains student information such as branch, age, attendance, and marks in Python, DBMS, and Mathematics.

import pandas as pd

data = {
    "Name": ["John", "Alice", "David", "Priya", "Rahul",
             "Sneha", "Arjun", "Kavya", "Rohan", "Meera"],

    "Branch": ["CSE", "ISE", "CSE", "ECE", "CSE",
               "ISE", "ECE", "CSE", "ISE", "ECE"],

    "Age": [19, 20, 21, 20, 22, 19, 21, 20, 22, 20],

    "Attendance": [85, 92, 78, 88, 75, 95, 82, 90, 70, 87],

    "Python": [78, 85, 90, 72, 76, 95, 80, 88, 68, 84],

    "DBMS": [82, 88, 85, 75, 79, 91, 83, 90, 70, 86],

    "Math": [75, 90, 88, 70, 74, 93, 81, 89, 65, 85]
}

df = pd.DataFrame(data)

print(df)

## 2. Project Objective

#The objectives of this project are:

#- Analyze student academic performance.
#- Calculate total and average marks.
#- Identify high and low performing students.
#- Analyze attendance and academic performance.
#- Compare performance across different subjects.
#- Visualize student performance using charts.
```
